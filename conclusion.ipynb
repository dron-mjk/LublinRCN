## Prawdopodne przyczyny błędów modelu

- W histogramie reszt (KDE) wskazano **prawostronną skośność** i długi dodatni ogon, czyli przypadki dużego **niedoszacowania cen**.
- Wykres **Q-Q** pokazuje odchylenie od normalności, szczególnie w prawym ogonie, co potwierdza **ekstremalne dodatnie błędy**.
- W interpretacji SHAP/LIME zapisano, że kluczowe są `lok_pow_uzyt` oraz cechy lokalizacyjne (ulice, dzielnice), ale nadal pojawiają się pojedyncze trudne obserwacje.

Podsumowanie: model działa najlepiej wśród testowanych, ale błędy koncentrują się w trudniejszych, skrajnych przypadkach, gdzie występują większe niedoszacowania.
