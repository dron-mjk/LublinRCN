# Przegląd literatury

Problem predykcji cen mieszkań jest ważny zarówno dla inwestorów, jak i dla miast. Najczęściej bada się go na podstawie cech mieszkania, lokalizacji i zmian cen w czasie.

W polskiej literaturze widać dwa główne podejścia. Pierwsze to modele hedoniczne i analiza zmian cen w czasie, często z elementem przestrzennym (np. Warszawa i duże miasta) {cite}`widlak2014` {cite}`bazyl2009`. Drugie podejście to uczenie maszynowe. Przykładem jest praca {cite}`galarowicz2024`, gdzie użyto modeli `Random Forest` i `XGBoost`.

W publikacjach zagranicznych często podkreśla się, że pomijanie aspektu przestrzennego pogarsza wyniki predykcji. Dlatego zaleca się łączenie modeli cen z danymi GIS i uwzględnianie zależności przestrzennych {cite}`aladwan2019` {cite}`wei2022` {cite}`cohen2020`.

W tej pracy łączę oba podejścia: modele ML i dane przestrzenne z geoportalu. Celem jest lepsza predykcja cen mieszkań oraz wykorzystanie nowoczesnych metod uczenia maszynowego
